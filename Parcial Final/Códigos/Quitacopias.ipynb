```python
import pandas as pd

# Cargar tu CSV
df = pd.read_csv("dataset_particulas_unificado_F.csv")

# Partículas a limpiar
particles_to_clean = ['electron', 'muon', 'tau', 'photon']

# Filtrar solo estas partículas
df_particles = df[df['particle_type'].isin(particles_to_clean)]

# Encontrar duplicados dentro del mismo evento:
# Se consideran duplicados si tienen exactamente igual:
# pt, eta, phi, particle_type, event_id
duplicates_mask = df_particles.duplicated(
    subset=['pt', 'eta', 'phi', 'particle_type', 'event_id'],
    keep='first'  # se mantiene la primera aparición
)

# Contar cuántos duplicados hay
num_duplicates = duplicates_mask.sum()

# Eliminar esos duplicados del dataframe original
df_clean = df.drop(df_particles[duplicates_mask].index)

# Mostrar cuántos duplicados fueron eliminados
print(f"Duplicados eliminados: {num_duplicates}")

# Si quieres guardarlo:
df_clean.to_csv("dataset_sin_repetidos.csv", index=False)
```

## CARGA 0 Y VACÍOS

```python
import pandas as pd
import numpy as np

# Cargar el archivo
df = pd.read_csv("dataset_ampliado.csv")

# =====================================
# 1) Contar valores vacíos por columna
#    (incluye strings vacíos " " o "")
# =====================================
df_temp = df.replace(r'^\s*$', np.nan, regex=True)  # detectar vacíos
missing_before = df_temp.isna().sum()

print("### Valores vacíos por columna (antes de limpiar) ###\n")
print(missing_before)
print("\nTotal de filas:", len(df))
print("-" * 60)

# =====================================
# 2) Normalizar vacíos → NaN
# =====================================
df = df_temp.copy()

# =====================================
# 3) charge vacío → 0
# =====================================
df['charge'] = pd.to_numeric(df['charge'], errors='coerce')
df['charge'] = df['charge'].fillna(0)

# =====================================
# 4) Eliminar filas con vacíos en otras columnas
# =====================================
columnas_excluir = ['charge']
columnas_revisar = [c for c in df.columns if c not in columnas_excluir]

df_limpio = df.dropna(subset=columnas_revisar)

# =====================================
# 5) Guardar resultado
# =====================================
df_limpio.to_csv("dataset_limpio.csv", index=False)

print("\n### Resumen de limpieza ###")
print("Filas originales:", len(df))
print("Filas finales:", len(df_limpio))
print("Filas eliminadas:", len(df) - len(df_limpio))

```
