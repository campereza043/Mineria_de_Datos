```python
import pandas as pd
import matplotlib.pyplot as plt

# Cargar archivo CSV
df = pd.read_csv("dataset_limpio.csv")
```

```python
conteo = df["particle_type"].value_counts()
print(conteo)
```

```python
plt.figure(figsize=(7,5))
conteo.plot(kind="bar")

plt.title("Cantidad de partículas por tipo")
plt.xlabel("Tipo de partícula")
plt.ylabel("Cantidad")
plt.xticks(rotation=45)
plt.grid(axis="y")

plt.show()
```
