```python
import pandas as pd
import glob
import os

ROOT = r"../Events/CSV"

# detección basada en strings
particle_labels = {
    "GsfElectrons": "electron",
    "TrackerMuons": "muon",
    "Photons": "photon",
    "Tracks": "track",
    "all_photons": "photon"
}

dfs = []

files = glob.glob(os.path.join(ROOT, "**/*.csv"), recursive=True)

print("\n### Archivos encontrados ###")
for f in files:
    print(" -", f)

print("\n### Procesando archivos ###")

# columnas mínimas obligatorias
required = ["pt", "eta", "phi"]

# columnas adicionales opcionales
optional_cols = [
    "charge",
    "pos_x", "pos_y", "pos_z"
]

for file in files:
    fname = os.path.basename(file)

    # Determinar tipo de partícula
    label = None
    for key in particle_labels:
        if key in fname:
            label = particle_labels[key]
            break

    if label is None:
        print(f" >> Saltando {fname} (tipo desconocido)")
        continue

    event_id = os.path.basename(os.path.dirname(file))

    event_id = event_id.replace("Event_", "")   # ← limpia el prefijo

    print(f"\nArchivo: {fname}")
    print(f"  Partícula detectada: {label}")
    print(f"  Evento: {event_id}")

    try:
        df = pd.read_csv(file)

        # ---------------------------
        # Normalización para fotones
        # ---------------------------
        if label == "photon" and "et" in df.columns and "pt" not in df.columns:
            df["pt"] = df["et"]

        # ---------------------------
        # Verificar columnas obligatorias
        # ---------------------------
        missing_req = [c for c in required if c not in df.columns]
        if missing_req:
            print("  !! Faltan columnas obligatorias:", missing_req)
            print("     Archivo ignorado.")
            continue

        # ---------------------------
        # Seleccionar columnas existentes
        # ---------------------------
        valid_optional = [c for c in optional_cols if c in df.columns]

        use_cols = required + valid_optional

        df = df[use_cols]

        # ---------------------------
        # Añadir etiquetas
        # ---------------------------
        df["particle_type"] = label
        df["event_id"] = event_id

        dfs.append(df)
        print(f"  OK — {len(df)} filas añadidas (columnas: {use_cols})")

    except Exception as e:
        print(f"  ERROR al leer {fname}: {e}")


# =============================================
# UNIR TODO
# =============================================
if not dfs:
    raise Exception("No se cargó ningún dataframe. Revisa las rutas.")

full_df = pd.concat(dfs, ignore_index=True)

print("\n### Conteo final de clases ###")
print(full_df["particle_type"].value_counts())

# Guardar dataset procesado
full_df.to_csv("dataset_ampliado.csv", index=False)

print("\nArchivo final guardado como dataset_ampliado.csv")
```
